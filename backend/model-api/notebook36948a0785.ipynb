```python
# ============================================================
# STEP 10 — CORRECT LEAKAGE-SAFE TRAIN / VALIDATION SPLIT
# ============================================================

from sklearn.model_selection import train_test_split


# ============================================================
# 1. CHECK HOW MANY UNIQUE TEXT GROUPS EACH CATEGORY HAS
# ============================================================

group_category_counts = (
    group_df["category"]
    .value_counts()
    .sort_index()
)

print("=" * 65)
print("UNIQUE TEXT GROUPS PER CATEGORY")
print("=" * 65)

print(group_category_counts)


# ============================================================
# 2. FIND CATEGORIES WITH ONLY ONE UNIQUE TEXT GROUP
# ============================================================

singleton_categories = (
    group_category_counts[
        group_category_counts < 2
    ]
    .index
    .tolist()
)

print("\nCategories with fewer than 2 unique text groups:")

if singleton_categories:
    print(singleton_categories)
else:
    print("None")


# ============================================================
# 3. SEPARATE SINGLETON CATEGORIES
# ============================================================

singleton_groups = group_df[
    group_df["category"].isin(
        singleton_categories
    )
].copy()


splittable_groups = group_df[
    ~group_df["category"].isin(
        singleton_categories
    )
].copy()


print("\n" + "=" * 65)
print("GROUP SPLITTING DATA")
print("=" * 65)

print(
    "Splittable text groups:",
    len(splittable_groups)
)

print(
    "Singleton text groups:",
    len(singleton_groups)
)


# ============================================================
# 4. STRATIFIED SPLIT FOR CATEGORIES WITH ENOUGH GROUPS
# ============================================================

train_groups, val_groups = train_test_split(
    splittable_groups["text"],
    test_size=0.15,
    random_state=42,
    stratify=splittable_groups["category"]
)


# ============================================================
# 5. FORCE SINGLETON GROUPS INTO TRAINING
# ============================================================

singleton_texts = set(
    singleton_groups["text"]
)


train_groups = list(train_groups) + list(
    singleton_texts
)

val_groups = list(val_groups)


# ============================================================
# 6. BUILD FINAL TRAIN / VALIDATION DATAFRAMES
# ============================================================

train_df = df_clean[
    df_clean["text"].isin(train_groups)
].copy()

val_df = df_clean[
    df_clean["text"].isin(val_groups)
].copy()


train_df = train_df.reset_index(
    drop=True
)

val_df = val_df.reset_index(
    drop=True
)


# ============================================================
# 7. CHECK DATA SIZES
# ============================================================

print("\n" + "=" * 65)
print("FINAL TRAIN / VALIDATION SPLIT")
print("=" * 65)

print(
    "Training rows:",
    len(train_df)
)

print(
    "Validation rows:",
    len(val_df)
)

print(
    "Total rows:",
    len(train_df) + len(val_df)
)


# ============================================================
# 8. CHECK TEXT LEAKAGE
# ============================================================

train_texts = set(
    train_df["text"]
)

val_texts = set(
    val_df["text"]
)

overlap = train_texts.intersection(
    val_texts
)

print("\nText groups appearing in BOTH sets:")

print(
    len(overlap)
)


# ============================================================
# 9. TRAIN CATEGORY DISTRIBUTION
# ============================================================

print("\n" + "=" * 65)
print("TRAIN CATEGORY DISTRIBUTION")
print("=" * 65)

print(
    train_df["category"]
    .value_counts()
)


# ============================================================
# 10. VALIDATION CATEGORY DISTRIBUTION
# ============================================================

print("\n" + "=" * 65)
print("VALIDATION CATEGORY DISTRIBUTION")
print("=" * 65)

print(
    val_df["category"]
    .value_counts()
)


# ============================================================
# 11. CHECK WHICH CATEGORIES ARE IN VALIDATION
# ============================================================

train_categories = set(
    train_df["category"]
)

val_categories = set(
    val_df["category"]
)

print("\n" + "=" * 65)
print("CATEGORY COVERAGE")
print("=" * 65)

print(
    "Categories in training:",
    len(train_categories)
)

print(
    "Categories in validation:",
    len(val_categories)
)

print(
    "Categories only in training:",
    sorted(
        train_categories - val_categories
    )
)

print(
    "Categories only in validation:",
    sorted(
        val_categories - train_categories
    )
)


# ============================================================
# 12. FINAL SANITY CHECK
# ============================================================

print("\n" + "=" * 65)
print("FINAL SANITY CHECK")
print("=" * 65)

print(
    "Conflicting text groups removed:",
    len(conflicting_texts)
)

print(
    "Train/validation text overlap:",
    len(overlap)
)

print(
    "Singleton categories forced into train:",
    singleton_categories
)
```

```python
# ============================================================
# STEP 11 — MuRIL TOKEN LENGTH ANALYSIS
# ============================================================

!pip -q install transformers

from transformers import AutoTokenizer
import numpy as np


# ============================================================
# 1. LOAD MuRIL TOKENIZER
# ============================================================

MODEL_NAME = "google/muril-base-cased"

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME
)

print("MuRIL tokenizer loaded successfully!")


# ============================================================
# 2. COMBINE TRAIN + VALIDATION TEXTS
# ============================================================

all_texts = pd.concat(
    [
        train_df["text"],
        val_df["text"]
    ],
    ignore_index=True
).astype(str)


print("\nTotal texts analyzed:")
print(len(all_texts))


# ============================================================
# 3. CALCULATE TOKEN LENGTHS
# ============================================================

token_lengths = []

for text in all_texts:

    tokens = tokenizer(
        text,
        truncation=False,
        add_special_tokens=True
    )

    token_lengths.append(
        len(tokens["input_ids"])
    )


token_lengths = np.array(
    token_lengths
)


# ============================================================
# 4. BASIC STATISTICS
# ============================================================

print("\n" + "=" * 65)
print("TOKEN LENGTH STATISTICS")
print("=" * 65)

print(
    "Minimum:",
    token_lengths.min()
)

print(
    "Maximum:",
    token_lengths.max()
)

print(
    "Mean:",
    round(
        token_lengths.mean(),
        2
    )
)

print(
    "Median:",
    np.median(token_lengths)
)

print(
    "95th percentile:",
    np.percentile(
        token_lengths,
        95
    )
)

print(
    "99th percentile:",
    np.percentile(
        token_lengths,
        99
    )
)


# ============================================================
# 5. CHECK COVERAGE FOR COMMON MAX_LENGTH VALUES
# ============================================================

print("\n" + "=" * 65)
print("MAX LENGTH COVERAGE")
print("=" * 65)

for max_len in [
    16,
    24,
    32,
    48,
    64
]:

    covered = (
        token_lengths <= max_len
    ).mean() * 100

    truncated = (
        token_lengths > max_len
    ).sum()

    print(
        f"MAX_LENGTH={max_len:2d}  | "
        f"Coverage={covered:6.2f}% | "
        f"Texts truncated={truncated}"
    )


# ============================================================
# 6. SHOW THE LONGEST TRANSACTIONS
# ============================================================

length_df = pd.DataFrame(
    {
        "text": all_texts,
        "token_length": token_lengths
    }
)

print("\n" + "=" * 65)
print("LONGEST TRANSACTION TEXTS")
print("=" * 65)

display(
    length_df
    .sort_values(
        "token_length",
        ascending=False
    )
    .head(20)
)


# ============================================================
# 7. SELECT A PRACTICAL MAX_LENGTH
# ============================================================

# We will choose after looking at the output.
```

```python
# ============================================================
# STEP 12 — CREATE PYTORCH DATASET AND DATALOADERS
# ============================================================

import torch
from torch.utils.data import Dataset, DataLoader


# ============================================================
# 1. SETTINGS
# ============================================================

MAX_LENGTH = 32
BATCH_SIZE = 16


# ============================================================
# 2. CREATE DATASET CLASS
# ============================================================

class TransactionDataset(Dataset):

    def __init__(
        self,
        dataframe,
        tokenizer,
        max_length=32
    ):

        self.data = dataframe.reset_index(
            drop=True
        )

        self.tokenizer = tokenizer
        self.max_length = max_length


    def __len__(self):

        return len(self.data)


    def __getitem__(self, idx):

        # ----------------------------------------------------
        # Get transaction text
        # ----------------------------------------------------

        text = str(
            self.data.loc[
                idx,
                "text"
            ]
        )


        # ----------------------------------------------------
        # Tokenize text using MuRIL tokenizer
        # ----------------------------------------------------

        encoding = self.tokenizer(
            text,
            truncation=True,
            padding="max_length",
            max_length=self.max_length,
            return_tensors="pt"
        )


        # Remove batch dimension
        input_ids = encoding[
            "input_ids"
        ].squeeze(0)

        attention_mask = encoding[
            "attention_mask"
        ].squeeze(0)


        # ----------------------------------------------------
        # Get numerical labels
        # ----------------------------------------------------

        type_label = torch.tensor(
            self.data.loc[
                idx,
                "type_id"
            ],
            dtype=torch.long
        )

        category_label = torch.tensor(
            self.data.loc[
                idx,
                "category_id"
            ],
            dtype=torch.long
        )


        # ----------------------------------------------------
        # Return everything needed by the model
        # ----------------------------------------------------

        return {
            "input_ids": input_ids,
            "attention_mask": attention_mask,
            "type_label": type_label,
            "category_label": category_label
        }


# ============================================================
# 3. CREATE TRAINING DATASET
# ============================================================

train_dataset = TransactionDataset(
    train_df,
    tokenizer,
    MAX_LENGTH
)


# ============================================================
# 4. CREATE VALIDATION DATASET
# ============================================================

val_dataset = TransactionDataset(
    val_df,
    tokenizer,
    MAX_LENGTH
)


print("=" * 65)
print("DATASET OBJECTS")
print("=" * 65)

print(
    "Training dataset size:",
    len(train_dataset)
)

print(
    "Validation dataset size:",
    len(val_dataset)
)


# ============================================================
# 5. SANITY CHECK ONE SAMPLE
# ============================================================

sample = train_dataset[0]

print("\n" + "=" * 65)
print("SAMPLE DATASET ITEM")
print("=" * 65)

print(
    "Input IDs shape:",
    sample["input_ids"].shape
)

print(
    "Attention mask shape:",
    sample["attention_mask"].shape
)

print(
    "Type label:",
    sample["type_label"].item()
)

print(
    "Category label:",
    sample["category_label"].item()
)


# ============================================================
# 6. CREATE TRAINING DATALOADER
# ============================================================

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=torch.cuda.is_available()
)


# ============================================================
# 7. CREATE VALIDATION DATALOADER
# ============================================================

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=torch.cuda.is_available()
)


# ============================================================
# 8. CHECK NUMBER OF BATCHES
# ============================================================

print("\n" + "=" * 65)
print("DATALOADERS")
print("=" * 65)

print(
    "Batch size:",
    BATCH_SIZE
)

print(
    "Training batches:",
    len(train_loader)
)

print(
    "Validation batches:",
    len(val_loader)
)


# ============================================================
# 9. CHECK ONE ACTUAL BATCH
# ============================================================

batch = next(
    iter(train_loader)
)

print("\n" + "=" * 65)
print("ONE TRAINING BATCH")
print("=" * 65)

print(
    "input_ids:",
    batch["input_ids"].shape
)

print(
    "attention_mask:",
    batch["attention_mask"].shape
)

print(
    "type_label:",
    batch["type_label"].shape
)

print(
    "category_label:",
    batch["category_label"].shape
)
```

```python
# ============================================================
# STEP 13 — LOAD MuRIL AND BUILD CLASSIFIER
# ============================================================

import torch
import torch.nn as nn

from transformers import AutoModel


# ============================================================
# 1. MODEL NAME
# ============================================================

MODEL_NAME = "google/muril-base-cased"


# ============================================================
# 2. LOAD MuRIL BACKBONE
# ============================================================

print("Loading MuRIL...")

bert = AutoModel.from_pretrained(
    MODEL_NAME
)

print("MuRIL loaded successfully!")

print(
    "\nHidden size:",
    bert.config.hidden_size
)


# ============================================================
# 3. CREATE CLASSIFIER
# ============================================================

class TransactionClassifier(nn.Module):

    def __init__(
        self,
        bert_model,
        num_types,
        num_categories
    ):

        super().__init__()

        # MuRIL backbone
        self.bert = bert_model

        # MuRIL-base hidden size = 768
        hidden = self.bert.config.hidden_size

        # Dropout helps reduce overfitting
        self.dropout = nn.Dropout(0.1)

        # Type classification head
        # 768 → 2
        self.type_head = nn.Linear(
            hidden,
            num_types
        )

        # Category classification head
        # 768 → 15
        self.category_head = nn.Linear(
            hidden,
            num_categories
        )


    def forward(
        self,
        input_ids,
        attention_mask
    ):

        # Pass text through MuRIL
        output = self.bert(
            input_ids=input_ids,
            attention_mask=attention_mask
        )

        # Take MuRIL pooled representation
        pooled = output.pooler_output

        # Apply dropout
        pooled = self.dropout(
            pooled
        )

        # Make two predictions
        type_logits = self.type_head(
            pooled
        )

        category_logits = self.category_head(
            pooled
        )

        return (
            type_logits,
            category_logits
        )


# ============================================================
# 4. CREATE MODEL
# ============================================================

model = TransactionClassifier(
    bert_model=bert,
    num_types=len(type_labels),
    num_categories=len(category_labels)
)


# ============================================================
# 5. SELECT DEVICE
# ============================================================

device = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

model = model.to(device)


print("\n" + "=" * 65)
print("MODEL INFORMATION")
print("=" * 65)

print(
    "Device:",
    device
)

print(
    "Number of types:",
    len(type_labels)
)

print(
    "Number of categories:",
    len(category_labels)
)


# ============================================================
# 6. COUNT PARAMETERS
# ============================================================

total_params = sum(
    p.numel()
    for p in model.parameters()
)

bert_params = sum(
    p.numel()
    for p in model.bert.parameters()
)

head_params = (
    total_params
    - bert_params
)


print(
    "\nMuRIL parameters:",
    f"{bert_params:,}"
)

print(
    "Classification head parameters:",
    f"{head_params:,}"
)

print(
    "Total parameters:",
    f"{total_params:,}"
)


# ============================================================
# 7. CHECK TRAINABLE PARAMETERS
# ============================================================

trainable_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

print(
    "Trainable parameters:",
    f"{trainable_params:,}"
)


# ============================================================
# 8. TEST ONE BATCH
# ============================================================

model.eval()

with torch.no_grad():

    batch = next(
        iter(train_loader)
    )

    input_ids = batch[
        "input_ids"
    ].to(device)

    attention_mask = batch[
        "attention_mask"
    ].to(device)

    type_logits, category_logits = model(
        input_ids=input_ids,
        attention_mask=attention_mask
    )


print("\n" + "=" * 65)
print("FORWARD PASS TEST")
print("=" * 65)

print(
    "Input shape:",
    input_ids.shape
)

print(
    "Type logits shape:",
    type_logits.shape
)

print(
    "Category logits shape:",
    category_logits.shape
)
```

```python
# ============================================================
# STEP 14 — CALCULATE CATEGORY CLASS WEIGHTS
# ============================================================

import numpy as np
import torch

from sklearn.utils.class_weight import compute_class_weight


# ============================================================
# 1. GET TRAINING CATEGORY LABELS
# ============================================================

train_category_ids = (
    train_df["category_id"]
    .to_numpy()
)


# ============================================================
# 2. CREATE ALL CATEGORY IDS
# ============================================================

all_category_ids = np.arange(
    len(category_labels)
)


# ============================================================
# 3. CALCULATE BALANCED CLASS WEIGHTS
# ============================================================

class_weights = compute_class_weight(
    class_weight="balanced",
    classes=all_category_ids,
    y=train_category_ids
)


# ============================================================
# 4. CONVERT TO PYTORCH TENSOR
# ============================================================

category_class_weights = torch.tensor(
    class_weights,
    dtype=torch.float32
).to(device)


# ============================================================
# 5. DISPLAY CATEGORY WEIGHTS
# ============================================================

print("=" * 65)
print("CATEGORY CLASS WEIGHTS")
print("=" * 65)

for category_id, category_name in enumerate(
    category_labels
):

    print(
        f"{category_id:2d} | "
        f"{category_name:15s} | "
        f"Train count = "
        f"{(train_category_ids == category_id).sum():3d} | "
        f"Weight = "
        f"{class_weights[category_id]:.4f}"
    )


# ============================================================
# 6. BASIC WEIGHT STATISTICS
# ============================================================

print("\n" + "=" * 65)
print("WEIGHT STATISTICS")
print("=" * 65)

print(
    "Minimum weight:",
    round(
        class_weights.min(),
        4
    )
)

print(
    "Maximum weight:",
    round(
        class_weights.max(),
        4
    )
)

print(
    "Mean weight:",
    round(
        class_weights.mean(),
        4
    )
)


# ============================================================
# 7. CHECK TENSOR
# ============================================================

print("\nPyTorch weight tensor:")

print(
    category_class_weights
)
```

```python
# ============================================================
# STEP 15 — LOSS FUNCTIONS + OPTIMIZER
# ============================================================

import torch
import torch.nn as nn


# ============================================================
# 1. TYPE LOSS
# ============================================================

# Type has only 2 classes:
# 0 = Expense
# 1 = Income

type_loss_fn = nn.CrossEntropyLoss()


# ============================================================
# 2. CATEGORY LOSS
# ============================================================

# Category loss uses our balanced class weights.
# This gives more importance to rare categories.

category_loss_fn = nn.CrossEntropyLoss(
    weight=category_class_weights
)


# ============================================================
# 3. OPTIMIZER
# ============================================================

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=2e-5
)


# ============================================================
# 4. DISPLAY SETTINGS
# ============================================================

print("=" * 65)
print("TRAINING CONFIGURATION")
print("=" * 65)

print(
    "Type loss:",
    "CrossEntropyLoss"
)

print(
    "Category loss:",
    "Weighted CrossEntropyLoss"
)

print(
    "Optimizer:",
    "AdamW"
)

print(
    "Learning rate:",
    2e-5
)


# ============================================================
# 5. TEST THE LOSS FUNCTIONS ON ONE BATCH
# ============================================================

model.train()

batch = next(
    iter(train_loader)
)

input_ids = batch[
    "input_ids"
].to(
    device,
    non_blocking=True
)

attention_mask = batch[
    "attention_mask"
].to(
    device,
    non_blocking=True
)

type_ids = batch[
    "type_label"
].to(
    device,
    non_blocking=True
)

category_ids = batch[
    "category_label"
].to(
    device,
    non_blocking=True
)


# ------------------------------------------------------------
# Clear previous gradients
# ------------------------------------------------------------

optimizer.zero_grad()


# ------------------------------------------------------------
# Forward pass
# ------------------------------------------------------------

type_logits, category_logits = model(
    input_ids,
    attention_mask
)


# ------------------------------------------------------------
# Calculate individual losses
# ------------------------------------------------------------

type_loss = type_loss_fn(
    type_logits,
    type_ids
)

category_loss = category_loss_fn(
    category_logits,
    category_ids
)


# ------------------------------------------------------------
# Combine losses
# ------------------------------------------------------------

total_loss = (
    type_loss
    + category_loss
)


# ============================================================
# 6. DISPLAY TEST RESULTS
# ============================================================

print("\n" + "=" * 65)
print("ONE BATCH LOSS TEST")
print("=" * 65)

print(
    "Type loss:",
    round(
        type_loss.item(),
        4
    )
)

print(
    "Category loss:",
    round(
        category_loss.item(),
        4
    )
)

print(
    "Total loss:",
    round(
        total_loss.item(),
        4
    )
)
```

```python
# ============================================================
# STEP 16 — TRAIN MuRIL
# ============================================================

import time
import torch


# ============================================================
# 1. TRAINING SETTINGS
# ============================================================

EPOCHS = 28

BEST_MODEL_PATH = (
    "/kaggle/working/"
    "muril_original_best.pt"
)

best_val_category_acc = 0.0


# ============================================================
# 2. TRAINING FUNCTION
# ============================================================

def train_one_epoch(
    model,
    loader,
    optimizer,
    device
):

    model.train()

    total_loss = 0.0

    correct_type = 0
    total_type = 0

    correct_category = 0
    total_category = 0


    for batch in loader:

        # ----------------------------------------------------
        # Move batch to GPU
        # ----------------------------------------------------

        input_ids = batch[
            "input_ids"
        ].to(
            device,
            non_blocking=True
        )

        attention_mask = batch[
            "attention_mask"
        ].to(
            device,
            non_blocking=True
        )

        type_ids = batch[
            "type_label"
        ].to(
            device,
            non_blocking=True
        )

        category_ids = batch[
            "category_label"
        ].to(
            device,
            non_blocking=True
        )


        # ----------------------------------------------------
        # Clear old gradients
        # ----------------------------------------------------

        optimizer.zero_grad()


        # ----------------------------------------------------
        # Forward pass
        # ----------------------------------------------------

        type_logits, category_logits = model(
            input_ids,
            attention_mask
        )


        # ----------------------------------------------------
        # Calculate losses
        # ----------------------------------------------------

        type_loss = type_loss_fn(
            type_logits,
            type_ids
        )

        category_loss = category_loss_fn(
            category_logits,
            category_ids
        )

        loss = (
            type_loss
            + category_loss
        )


        # ----------------------------------------------------
        # Backpropagation
        # ----------------------------------------------------

        loss.backward()


        # ----------------------------------------------------
        # Update model
        # ----------------------------------------------------

        optimizer.step()


        # ----------------------------------------------------
        # Store loss
        # ----------------------------------------------------

        total_loss += loss.item()


        # ----------------------------------------------------
        # Type predictions
        # ----------------------------------------------------

        type_predictions = torch.argmax(
            type_logits,
            dim=1
        )


        # ----------------------------------------------------
        # Category predictions
        # ----------------------------------------------------

        category_predictions = torch.argmax(
            category_logits,
            dim=1
        )


        # ----------------------------------------------------
        # Count correct predictions
        # ----------------------------------------------------

        correct_type += (
            type_predictions == type_ids
        ).sum().item()

        total_type += (
            type_ids.size(0)
        )


        correct_category += (
            category_predictions == category_ids
        ).sum().item()

        total_category += (
            category_ids.size(0)
        )


    # ========================================================
    # EPOCH METRICS
    # ========================================================

    average_loss = (
        total_loss
        / len(loader)
    )

    type_accuracy = (
        correct_type
        / total_type
        * 100
    )

    category_accuracy = (
        correct_category
        / total_category
        * 100
    )

    return (
        average_loss,
        type_accuracy,
        category_accuracy
    )


# ============================================================
# 3. VALIDATION FUNCTION
# ============================================================

def validate(
    model,
    loader,
    device
):

    model.eval()

    total_loss = 0.0

    correct_type = 0
    total_type = 0

    correct_category = 0
    total_category = 0


    with torch.no_grad():

        for batch in loader:

            # ------------------------------------------------
            # Move batch to GPU
            # ------------------------------------------------

            input_ids = batch[
                "input_ids"
            ].to(
                device,
                non_blocking=True
            )

            attention_mask = batch[
                "attention_mask"
            ].to(
                device,
                non_blocking=True
            )

            type_ids = batch[
                "type_label"
            ].to(
                device,
                non_blocking=True
            )

            category_ids = batch[
                "category_label"
            ].to(
                device,
                non_blocking=True
            )


            # ------------------------------------------------
            # Forward pass
            # ------------------------------------------------

            type_logits, category_logits = model(
                input_ids,
                attention_mask
            )


            # ------------------------------------------------
            # Validation loss
            # ------------------------------------------------

            type_loss = type_loss_fn(
                type_logits,
                type_ids
            )

            category_loss = category_loss_fn(
                category_logits,
                category_ids
            )

            loss = (
                type_loss
                + category_loss
            )

            total_loss += loss.item()


            # ------------------------------------------------
            # Predictions
            # ------------------------------------------------

            type_predictions = torch.argmax(
                type_logits,
                dim=1
            )

            category_predictions = torch.argmax(
                category_logits,
                dim=1
            )


            # ------------------------------------------------
            # Accuracy
            # ------------------------------------------------

            correct_type += (
                type_predictions == type_ids
            ).sum().item()

            total_type += (
                type_ids.size(0)
            )

            correct_category += (
                category_predictions == category_ids
            ).sum().item()

            total_category += (
                category_ids.size(0)
            )


    average_loss = (
        total_loss
        / len(loader)
    )

    type_accuracy = (
        correct_type
        / total_type
        * 100
    )

    category_accuracy = (
        correct_category
        / total_category
        * 100
    )

    return (
        average_loss,
        type_accuracy,
        category_accuracy
    )


# ============================================================
# 4. TRAINING LOOP
# ============================================================

print("=" * 70)
print("STARTING MuRIL TRAINING")
print("=" * 70)

print(
    "Epochs:",
    EPOCHS
)

print(
    "Best model:",
    BEST_MODEL_PATH
)

print()


for epoch in range(EPOCHS):

    start_time = time.time()


    # --------------------------------------------------------
    # Training
    # --------------------------------------------------------

    train_loss, train_type_acc, train_category_acc = (
        train_one_epoch(
            model,
            train_loader,
            optimizer,
            device
        )
    )


    # --------------------------------------------------------
    # Validation
    # --------------------------------------------------------

    val_loss, val_type_acc, val_category_acc = (
        validate(
            model,
            val_loader,
            device
        )
    )


    # --------------------------------------------------------
    # Time
    # --------------------------------------------------------

    epoch_time = (
        time.time()
        - start_time
    )


    # --------------------------------------------------------
    # Print results
    # --------------------------------------------------------

    print(
        f"Epoch {epoch + 1:02d}/{EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Type Acc: {train_type_acc:.2f}% | "
        f"Train Cat Acc: {train_category_acc:.2f}% | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Type Acc: {val_type_acc:.2f}% | "
        f"Val Cat Acc: {val_category_acc:.2f}% | "
        f"Time: {epoch_time:.1f}s"
    )


    # --------------------------------------------------------
    # Save best model
    # --------------------------------------------------------

    if val_category_acc > best_val_category_acc:

        best_val_category_acc = (
            val_category_acc
        )

        torch.save(
            model.state_dict(),
            BEST_MODEL_PATH
        )

        print(
            f"  --> NEW BEST MODEL SAVED "
            f"(Val Cat Acc: "
            f"{best_val_category_acc:.2f}%)"
        )

    else:

        print(
            f"  --> No improvement "
            f"(Best: "
            f"{best_val_category_acc:.2f}%)"
        )


# ============================================================
# 5. TRAINING COMPLETE
# ============================================================

print("\n" + "=" * 70)
print("TRAINING COMPLETE")
print("=" * 70)

print(
    "Best validation category accuracy:",
    f"{best_val_category_acc:.2f}%"
)

print(
    "Best model saved at:",
    BEST_MODEL_PATH
)
```

```python
# ============================================================
# STEP 21 — CREATE WHATSAPP / HINGLISH DATASET
# ============================================================

import pandas as pd


# ============================================================
# 1. WHATSAPP MESSAGE TEMPLATES
# ============================================================

expense_templates = [
    "yaar {text} pe paise lag gaye",
    "bhai {text} ka kharcha hua",
    "aaj {text} ke liye payment kiya",
    "maine {text} liya, paise chale gaye",
    "aaj {text} ke paise diye",
    "just {text} ka payment kar diya",
    "{text} le liya yaar",
    "bhai aaj {text} par spend kiya",
    "{text} ka paisa de diya",
    "aaj {text} ke liye pay kiya"
]

income_templates = [
    "yaar {text} aa gayi",
    "bhai {text} mil gayi",
    "aaj {text} receive hui",
    "account me {text} aayi",
    "{text} credited ho gayi",
    "aaj mujhe {text} mila",
    "{text} aa gaya account me",
    "bhai aaj {text} receive hua"
]


# ============================================================
# 2. FUNCTION TO GENERATE WHATSAPP VARIANTS
# ============================================================

def generate_whatsapp_variants(
    dataframe,
    variants_per_text=8
):

    rows = []

    # Use unique text + category + type
    unique_data = (
        dataframe[
            [
                "text",
                "category",
                "type",
                "category_id",
                "type_id"
            ]
        ]
        .drop_duplicates()
        .reset_index(drop=True)
    )

    for _, row in unique_data.iterrows():

        original_text = str(
            row["text"]
        )

        category = row["category"]
        transaction_type = row["type"]

        category_id = row["category_id"]
        type_id = row["type_id"]

        # ----------------------------------------------------
        # Choose templates based on income/expense
        # ----------------------------------------------------

        if transaction_type == "Income":
            templates = income_templates
        else:
            templates = expense_templates

        # ----------------------------------------------------
        # Keep only requested number
        # ----------------------------------------------------

        selected_templates = templates[
            :variants_per_text
        ]

        # ----------------------------------------------------
        # Add original text
        # ----------------------------------------------------

        rows.append({
            "text": original_text,
            "category": category,
            "type": transaction_type,
            "category_id": category_id,
            "type_id": type_id,
            "style": "original"
        })

        # ----------------------------------------------------
        # Add WhatsApp variants
        # ----------------------------------------------------

        for template in selected_templates:

            whatsapp_text = template.format(
                text=original_text
            )

            rows.append({
                "text": whatsapp_text,
                "category": category,
                "type": transaction_type,
                "category_id": category_id,
                "type_id": type_id,
                "style": "whatsapp"
            })

    return pd.DataFrame(rows)


# ============================================================
# 3. GENERATE TRAINING DATA
# ============================================================

whatsapp_train_df = generate_whatsapp_variants(
    train_df,
    variants_per_text=8
)


# ============================================================
# 4. GENERATE VALIDATION DATA
# ============================================================

whatsapp_val_df = generate_whatsapp_variants(
    val_df,
    variants_per_text=8
)


# ============================================================
# 5. DISPLAY DATASET SIZES
# ============================================================

print("=" * 70)
print("WHATSAPP DATASET SIZES")
print("=" * 70)

print(
    "Original training rows:",
    len(train_df)
)

print(
    "WhatsApp training rows:",
    len(whatsapp_train_df)
)

print(
    "Original validation rows:",
    len(val_df)
)

print(
    "WhatsApp validation rows:",
    len(whatsapp_val_df)
)


# ============================================================
# 6. CHECK CATEGORY DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("WHATSAPP TRAIN CATEGORY DISTRIBUTION")
print("=" * 70)

print(
    whatsapp_train_df[
        "category"
    ].value_counts()
)


print("\n" + "=" * 70)
print("WHATSAPP VALIDATION CATEGORY DISTRIBUTION")
print("=" * 70)

print(
    whatsapp_val_df[
        "category"
    ].value_counts()
)


# ============================================================
# 7. SHOW ACTUAL WHATSAPP EXAMPLES
# ============================================================

print("\n" + "=" * 70)
print("WHATSAPP-STYLE EXAMPLES")
print("=" * 70)

display(
    whatsapp_train_df[
        whatsapp_train_df["style"] == "whatsapp"
    ][
        [
            "text",
            "category"
        ]
    ].head(40)
)


# ============================================================
# 8. CHECK FOR CROSS-SPLIT TEXT LEAKAGE
# ============================================================

train_texts = set(
    whatsapp_train_df["text"]
)

val_texts = set(
    whatsapp_val_df["text"]
)

overlap = train_texts.intersection(
    val_texts
)

print("\n" + "=" * 70)
print("CROSS-SPLIT LEAKAGE CHECK")
print("=" * 70)

print(
    "Exact texts appearing in both:",
    len(overlap)
)


# ============================================================
# 9. SHOW SOME REALISTIC EXAMPLES FROM DIFFERENT CLASSES
# ============================================================

print("\n" + "=" * 70)
print("SAMPLE BY CATEGORY")
print("=" * 70)

for category in [
    "Food",
    "Apparel",
    "Health",
    "Transportation",
    "Subscription",
    "Family",
    "Leisure"
]:

    sample = whatsapp_train_df[
        (
            whatsapp_train_df["category"]
            == category
        )
        &
        (
            whatsapp_train_df["style"]
            == "whatsapp"
        )
    ].head(3)

    print("\nCATEGORY:", category)

    for text in sample["text"]:
        print("-", text)
```

```python
# ============================================================
# STEP 23 — BUILD WHATSAPP TRAINING / VALIDATION LOADERS
# ============================================================

import torch
from torch.utils.data import Dataset, DataLoader


# ============================================================
# 1. SETTINGS
# ============================================================

WHATSAPP_MAX_LENGTH = 32
WHATSAPP_BATCH_SIZE = 16


# ============================================================
# 2. DATASET CLASS
# ============================================================

class WhatsAppTransactionDataset(Dataset):

    def __init__(
        self,
        dataframe,
        tokenizer,
        max_length=32
    ):

        self.data = dataframe.reset_index(
            drop=True
        )

        self.tokenizer = tokenizer
        self.max_length = max_length


    def __len__(self):

        return len(self.data)


    def __getitem__(self, idx):

        text = str(
            self.data.loc[idx, "text"]
        )

        encoding = self.tokenizer(
            text,
            truncation=True,
            padding="max_length",
            max_length=self.max_length,
            return_tensors="pt"
        )

        return {
            "input_ids": encoding[
                "input_ids"
            ].squeeze(0),

            "attention_mask": encoding[
                "attention_mask"
            ].squeeze(0),

            "type_label": torch.tensor(
                self.data.loc[idx, "type_id"],
                dtype=torch.long
            ),

            "category_label": torch.tensor(
                self.data.loc[idx, "category_id"],
                dtype=torch.long
            )
        }


# ============================================================
# 3. CREATE DATASETS
# ============================================================

whatsapp_train_dataset = WhatsAppTransactionDataset(
    whatsapp_train_df,
    tokenizer,
    WHATSAPP_MAX_LENGTH
)

whatsapp_val_dataset = WhatsAppTransactionDataset(
    whatsapp_val_df,
    tokenizer,
    WHATSAPP_MAX_LENGTH
)


# ============================================================
# 4. CREATE DATALOADERS
# ============================================================

whatsapp_train_loader = DataLoader(
    whatsapp_train_dataset,
    batch_size=WHATSAPP_BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=torch.cuda.is_available()
)

whatsapp_val_loader = DataLoader(
    whatsapp_val_dataset,
    batch_size=WHATSAPP_BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=torch.cuda.is_available()
)


# ============================================================
# 5. DISPLAY INFORMATION
# ============================================================

print("=" * 70)
print("WHATSAPP DATALOADERS")
print("=" * 70)

print(
    "Training dataset:",
    len(whatsapp_train_dataset)
)

print(
    "Validation dataset:",
    len(whatsapp_val_dataset)
)

print(
    "Batch size:",
    WHATSAPP_BATCH_SIZE
)

print(
    "Training batches:",
    len(whatsapp_train_loader)
)

print(
    "Validation batches:",
    len(whatsapp_val_loader)
)


# ============================================================
# 6. CHECK ONE BATCH
# ============================================================

batch = next(
    iter(whatsapp_train_loader)
)

print("\n" + "=" * 70)
print("ONE WHATSAPP BATCH")
print("=" * 70)

print(
    "input_ids:",
    batch["input_ids"].shape
)

print(
    "attention_mask:",
    batch["attention_mask"].shape
)

print(
    "type_label:",
    batch["type_label"].shape
)

print(
    "category_label:",
    batch["category_label"].shape
)
```

```python
# ============================================================
# STEP 25 — WHATSAPP FINE-TUNING
# ============================================================

import torch
import numpy as np

from sklearn.metrics import (
    accuracy_score,
    f1_score
)


# ============================================================
# 1. CONFIGURATION
# ============================================================

NUM_EPOCHS = 6

BEST_WHATSAPP_MODEL_PATH = (
    "/kaggle/working/"
    "muril_whatsapp_best.pt"
)

best_val_macro_f1 = -1.0
best_val_accuracy = -1.0
patience = 2
epochs_without_improvement = 0


# ============================================================
# 2. EVALUATION FUNCTION
# ============================================================

def evaluate_whatsapp(loader):

    model.eval()

    all_type_true = []
    all_type_pred = []

    all_category_true = []
    all_category_pred = []

    total_loss = 0.0

    with torch.no_grad():

        for batch in loader:

            input_ids = batch[
                "input_ids"
            ].to(
                device,
                non_blocking=True
            )

            attention_mask = batch[
                "attention_mask"
            ].to(
                device,
                non_blocking=True
            )

            type_ids = batch[
                "type_label"
            ].to(
                device,
                non_blocking=True
            )

            category_ids = batch[
                "category_label"
            ].to(
                device,
                non_blocking=True
            )

            # Forward pass
            type_logits, category_logits = model(
                input_ids,
                attention_mask
            )

            # Loss
            type_loss = whatsapp_type_loss_fn(
                type_logits,
                type_ids
            )

            category_loss = whatsapp_category_loss_fn(
                category_logits,
                category_ids
            )

            loss = type_loss + category_loss

            total_loss += loss.item()

            # Predictions
            type_preds = torch.argmax(
                type_logits,
                dim=1
            )

            category_preds = torch.argmax(
                category_logits,
                dim=1
            )

            # Move to CPU
            all_type_true.extend(
                type_ids.cpu().numpy()
            )

            all_type_pred.extend(
                type_preds.cpu().numpy()
            )

            all_category_true.extend(
                category_ids.cpu().numpy()
            )

            all_category_pred.extend(
                category_preds.cpu().numpy()
            )

    # --------------------------------------------------------
    # Convert to NumPy
    # --------------------------------------------------------

    all_type_true = np.array(
        all_type_true
    )

    all_type_pred = np.array(
        all_type_pred
    )

    all_category_true = np.array(
        all_category_true
    )

    all_category_pred = np.array(
        all_category_pred
    )

    # --------------------------------------------------------
    # Type metrics
    # --------------------------------------------------------

    type_accuracy = accuracy_score(
        all_type_true,
        all_type_pred
    )

    type_macro_f1 = f1_score(
        all_type_true,
        all_type_pred,
        average="macro",
        zero_division=0
    )

    # --------------------------------------------------------
    # Category metrics
    # --------------------------------------------------------

    category_accuracy = accuracy_score(
        all_category_true,
        all_category_pred
    )

    # Only categories actually present in validation
    present_labels = np.unique(
        all_category_true
    )

    category_macro_f1 = f1_score(
        all_category_true,
        all_category_pred,
        labels=present_labels,
        average="macro",
        zero_division=0
    )

    average_loss = (
        total_loss / len(loader)
    )

    return {
        "loss": average_loss,
        "type_accuracy": type_accuracy,
        "type_macro_f1": type_macro_f1,
        "category_accuracy": category_accuracy,
        "category_macro_f1": category_macro_f1,
        "type_true": all_type_true,
        "type_pred": all_type_pred,
        "category_true": all_category_true,
        "category_pred": all_category_pred
    }


# ============================================================
# 3. TRAINING LOOP
# ============================================================

print("=" * 70)
print("STARTING WHATSAPP MÚRIL FINE-TUNING")
print("=" * 70)

for epoch in range(NUM_EPOCHS):

    model.train()

    running_loss = 0.0

    train_type_true = []
    train_type_pred = []

    train_category_true = []
    train_category_pred = []

    for batch in whatsapp_train_loader:

        # ----------------------------------------------------
        # Move batch to GPU / device
        # ----------------------------------------------------

        input_ids = batch[
            "input_ids"
        ].to(
            device,
            non_blocking=True
        )

        attention_mask = batch[
            "attention_mask"
        ].to(
            device,
            non_blocking=True
        )

        type_ids = batch[
            "type_label"
        ].to(
            device,
            non_blocking=True
        )

        category_ids = batch[
            "category_label"
        ].to(
            device,
            non_blocking=True
        )

        # ----------------------------------------------------
        # Clear old gradients
        # ----------------------------------------------------

        whatsapp_optimizer.zero_grad(
            set_to_none=True
        )

        # ----------------------------------------------------
        # Forward pass
        # ----------------------------------------------------

        type_logits, category_logits = model(
            input_ids,
            attention_mask
        )

        # ----------------------------------------------------
        # Calculate losses
        # ----------------------------------------------------

        type_loss = whatsapp_type_loss_fn(
            type_logits,
            type_ids
        )

        category_loss = whatsapp_category_loss_fn(
            category_logits,
            category_ids
        )

        total_loss = (
            type_loss
            + category_loss
        )

        # ----------------------------------------------------
        # Backpropagation
        # ----------------------------------------------------

        total_loss.backward()

        # ----------------------------------------------------
        # Gradient clipping
        # ----------------------------------------------------

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        # ----------------------------------------------------
        # Update weights
        # ----------------------------------------------------

        whatsapp_optimizer.step()

        running_loss += total_loss.item()

        # ----------------------------------------------------
        # Save predictions for training metrics
        # ----------------------------------------------------

        type_preds = torch.argmax(
            type_logits,
            dim=1
        )

        category_preds = torch.argmax(
            category_logits,
            dim=1
        )

        train_type_true.extend(
            type_ids.detach().cpu().numpy()
        )

        train_type_pred.extend(
            type_preds.detach().cpu().numpy()
        )

        train_category_true.extend(
            category_ids.detach().cpu().numpy()
        )

        train_category_pred.extend(
            category_preds.detach().cpu().numpy()
        )

    # ========================================================
    # TRAINING METRICS
    # ========================================================

    train_loss = (
        running_loss /
        len(whatsapp_train_loader)
    )

    train_type_accuracy = accuracy_score(
        train_type_true,
        train_type_pred
    )

    train_category_accuracy = accuracy_score(
        train_category_true,
        train_category_pred
    )

    # ========================================================
    # VALIDATION
    # ========================================================

    val_metrics = evaluate_whatsapp(
        whatsapp_val_loader
    )

    # ========================================================
    # DISPLAY RESULTS
    # ========================================================

    print("\n" + "-" * 70)

    print(
        f"Epoch {epoch + 1}/{NUM_EPOCHS}"
    )

    print(
        f"Train Loss: "
        f"{train_loss:.4f}"
    )

    print(
        f"Train Type Acc: "
        f"{train_type_accuracy * 100:.2f}%"
    )

    print(
        f"Train Category Acc: "
        f"{train_category_accuracy * 100:.2f}%"
    )

    print(
        f"Val Loss: "
        f"{val_metrics['loss']:.4f}"
    )

    print(
        f"Val Type Acc: "
        f"{val_metrics['type_accuracy'] * 100:.2f}%"
    )

    print(
        f"Val Type Macro-F1: "
        f"{val_metrics['type_macro_f1']:.4f}"
    )

    print(
        f"Val Category Acc: "
        f"{val_metrics['category_accuracy'] * 100:.2f}%"
    )

    print(
        f"Val Category Macro-F1: "
        f"{val_metrics['category_macro_f1']:.4f}"
    )

    # ========================================================
    # SAVE BEST MODEL
    # ========================================================

    current_f1 = (
        val_metrics["category_macro_f1"]
    )

    if current_f1 > best_val_macro_f1:

        best_val_macro_f1 = current_f1

        best_val_accuracy = (
            val_metrics["category_accuracy"]
        )

        epochs_without_improvement = 0

        torch.save(
            model.state_dict(),
            BEST_WHATSAPP_MODEL_PATH
        )

        print(
            "\n✅ NEW BEST MODEL SAVED"
        )

        print(
            f"Best Val Category Macro-F1: "
            f"{best_val_macro_f1:.4f}"
        )

        print(
            f"Best Val Category Accuracy: "
            f"{best_val_accuracy * 100:.2f}%"
        )

    else:

        epochs_without_improvement += 1

        print(
            f"\nNo improvement."
            f" Patience: "
            f"{epochs_without_improvement}/{patience}"
        )

    # ========================================================
    # EARLY STOPPING
    # ========================================================

    if epochs_without_improvement >= patience:

        print(
            "\n⏹ Early stopping triggered."
        )

        break


# ============================================================
# 4. LOAD BEST WHATSAPP CHECKPOINT
# ============================================================

print("\n" + "=" * 70)
print("LOADING BEST WHATSAPP CHECKPOINT")
print("=" * 70)

model.load_state_dict(
    torch.load(
        BEST_WHATSAPP_MODEL_PATH,
        map_location=device
    )
)

model = model.to(device)

print(
    "Best WhatsApp model loaded successfully!"
)

print(
    f"Best validation category accuracy: "
    f"{best_val_accuracy * 100:.2f}%"
)

print(
    f"Best validation category macro-F1: "
    f"{best_val_macro_f1:.4f}"
)
```

```python
# ============================================================
# STEP 29 — FINAL HYBRID PARSER
# ============================================================

import re
import torch


# ============================================================
# 1. ADD GENERIC "OTHER" KEYWORDS
# ============================================================

CATEGORY_KEYWORDS["Other"] = [

    "random expense",
    "random kharcha",
    "miscellaneous",
    "miscellaneous expense",
    "misc expense",
    "misc kharcha",
    "other expense",
    "other kharcha",
    "random kharch",
    "miscellaneous kharch",
    "misc",
    "unplanned expense",
    "unplanned kharcha"
]


# ============================================================
# 2. IMPROVED CATEGORY RULE FUNCTION
# ============================================================

def rule_based_category(text):

    text_lower = text.lower()

    matches = []

    for category, keywords in CATEGORY_KEYWORDS.items():

        for keyword in keywords:

            if keyword in text_lower:

                matches.append(
                    (
                        category,
                        len(keyword)
                    )
                )

    if not matches:
        return None

    # Prefer the most specific / longest keyword
    matches.sort(
        key=lambda x: x[1],
        reverse=True
    )

    return matches[0][0]


# ============================================================
# 3. MÚRIL FALLBACK CLASSIFIER
# ============================================================

def muril_category_prediction(text):

    model.eval()

    encoded = tokenizer(
        text,
        padding="max_length",
        truncation=True,
        max_length=32,
        return_tensors="pt"
    )

    input_ids = encoded[
        "input_ids"
    ].to(device)

    attention_mask = encoded[
        "attention_mask"
    ].to(device)

    with torch.no_grad():

        type_logits, category_logits = model(
            input_ids,
            attention_mask
        )

        probabilities = torch.softmax(
            category_logits,
            dim=1
        )

        confidence, prediction = torch.max(
            probabilities,
            dim=1
        )

    predicted_category = category_labels[
        prediction.item()
    ]

    confidence_value = confidence.item()

    return (
        predicted_category,
        confidence_value
    )


# ============================================================
# 4. FINAL TRANSACTION PARSER
# ============================================================

def parse_whatsapp_message_final(text):

    # --------------------------------------------------------
    # Amount
    # --------------------------------------------------------

    amount = extract_amount(text)

    # --------------------------------------------------------
    # Type
    # --------------------------------------------------------

    transaction_type = rule_based_type(text)

    # --------------------------------------------------------
    # Rule-based category
    # --------------------------------------------------------

    category = rule_based_category(text)

    source = "rule"
    confidence = 1.0

    # --------------------------------------------------------
    # MuRIL fallback
    # --------------------------------------------------------

    if category is None:

        category, confidence = (
            muril_category_prediction(text)
        )

        source = "muril"

    return {

        "text": text,

        "amount": amount,

        "category": category,

        "type": transaction_type,

        "category_source": source,

        "category_confidence": confidence
    }


# ============================================================
# 5. TEST THE TWO PREVIOUS FAILURES
# ============================================================

test_messages = [

    "yaar 250 ka random expense ho gaya",

    "500 ka miscellaneous kharcha ho gaya",

    "kal kuch ajeeb sa kharcha hua"
]


print("=" * 70)
print("FINAL HYBRID PARSER TEST")
print("=" * 70)


for message in test_messages:

    result = parse_whatsapp_message_final(
        message
    )

    print("\nMessage:", message)

    print(
        "Amount:",
        result["amount"]
    )

    print(
        "Category:",
        result["category"]
    )

    print(
        "Type:",
        result["type"]
    )

    print(
        "Source:",
        result["category_source"]
    )

    print(
        "Confidence:",
        round(
            result["category_confidence"],
            4
        )
    )
```

```python
# ============================================================
# STEP 30 — PRODUCTION-SAFE WHATSAPP PARSER
# ============================================================

import json
import torch


# ============================================================
# 1. CONFIDENCE THRESHOLD
# ============================================================

MURIL_CONFIDENCE_THRESHOLD = 0.95


# ============================================================
# 2. SAFE FINAL PARSER
# ============================================================

def parse_whatsapp_message_production(text):

    # --------------------------------------------------------
    # Extract amount
    # --------------------------------------------------------

    amount = extract_amount(text)

    # --------------------------------------------------------
    # Detect transaction type
    # --------------------------------------------------------

    transaction_type = rule_based_type(text)

    # --------------------------------------------------------
    # First try deterministic rules
    # --------------------------------------------------------

    rule_category = rule_based_category(text)

    category_source = None
    category_confidence = None
    category = None

    # --------------------------------------------------------
    # RULE MATCH
    # --------------------------------------------------------

    if rule_category is not None:

        category = rule_category

        category_source = "rule"

        category_confidence = 1.0

    # --------------------------------------------------------
    # MÚRIL FALLBACK
    # --------------------------------------------------------

    else:

        muril_category, muril_confidence = (
            muril_category_prediction(text)
        )

        if (
            muril_confidence
            >=
            MURIL_CONFIDENCE_THRESHOLD
        ):

            category = muril_category

            category_source = "muril"

            category_confidence = (
                muril_confidence
            )

        else:

            category = None

            category_source = "review"

            category_confidence = (
                muril_confidence
            )

    # --------------------------------------------------------
    # DETERMINE WHETHER TRANSACTION IS SAFE TO SAVE
    # --------------------------------------------------------

    ready_to_save = (

        amount is not None

        and

        category is not None

        and

        transaction_type is not None
    )

    # --------------------------------------------------------
    # REVIEW REASON
    # --------------------------------------------------------

    review_reason = None

    if amount is None:

        review_reason = (
            "Amount could not be extracted"
        )

    elif category is None:

        review_reason = (
            "Category confidence is below threshold"
        )

    # --------------------------------------------------------
    # FINAL STRUCTURED RESULT
    # --------------------------------------------------------

    result = {

        "text": text,

        "amount": amount,

        "category": category,

        "type": transaction_type,

        "category_source": category_source,

        "category_confidence":
            round(
                category_confidence,
                4
            )
            if category_confidence is not None
            else None,

        "ready_to_save":
            ready_to_save,

        "review_reason":
            review_reason
    }

    return result


# ============================================================
# 3. TEST MESSAGES
# ============================================================

test_messages = [

    "yaar 250 ka pizza mangaya",

    "amazon se 1800 ke shoes liye",

    "doctor ko 600 diye",

    "uber se ghar aaya 220",

    "aaj salary 45000 aa gayi",

    "yaar 250 ka random expense ho gaya",

    "500 ka miscellaneous kharcha ho gaya",

    "kal kuch ajeeb sa kharcha hua"
]


# ============================================================
# 4. DISPLAY RESULTS
# ============================================================

print("=" * 70)
print("PRODUCTION-SAFE WHATSAPP PARSER")
print("=" * 70)

for message in test_messages:

    result = parse_whatsapp_message_production(
        message
    )

    print("\nMessage:")
    print(message)

    print("\nStructured result:")

    print(
        json.dumps(
            result,
            indent=2,
            ensure_ascii=False
        )
    )
```

```python
# ============================================================
# STEP 31 — DATABASE-READY TRANSACTION JSON
# ============================================================

import json


def create_database_payload(text):

    result = parse_whatsapp_message_production(text)

    # --------------------------------------------------------
    # CASE 1: SAFE TO SAVE
    # --------------------------------------------------------

    if result["ready_to_save"]:

        payload = {
            "amount": result["amount"],
            "category": result["category"],
            "type": result["type"],
            "description": result["text"]
        }

        return {
            "status": "success",
            "save_to_database": True,
            "transaction": payload
        }

    # --------------------------------------------------------
    # CASE 2: NEEDS REVIEW
    # --------------------------------------------------------

    return {
        "status": "needs_review",
        "save_to_database": False,
        "transaction": None,
        "message": (
            "I couldn't understand this transaction "
            "completely. Please provide the amount "
            "and a little more detail."
        )
    }


# ============================================================
# TEST
# ============================================================

messages = [

    "yaar 250 ka pizza mangaya",

    "amazon se 1800 ke shoes liye",

    "doctor ko 600 diye",

    "uber se ghar aaya 220",

    "aaj salary 45000 aa gayi",

    "500 ka miscellaneous kharcha ho gaya",

    "kal kuch ajeeb sa kharcha hua"
]


print("=" * 70)
print("DATABASE-READY WHATSAPP PAYLOADS")
print("=" * 70)

for message in messages:

    print("\nIncoming WhatsApp message:")
    print(message)

    payload = create_database_payload(
        message
    )

    print("\nBackend payload:")

    print(
        json.dumps(
            payload,
            indent=2,
            ensure_ascii=False
        )
    )
```

```python


# ============================================================
# STEP 32 — CREATE API VERSION OF THE PARSER
# ============================================================

from fastapi import FastAPI
from pydantic import BaseModel
import json


# ============================================================
# 1. CREATE FASTAPI APP
# ============================================================

app = FastAPI(
    title="WhatsApp Finance Parser",
    description="AI-powered WhatsApp transaction parser",
    version="1.0.0"
)


# ============================================================
# 2. REQUEST FORMAT
# ============================================================

class TransactionRequest(BaseModel):

    message: str


# ============================================================
# 3. HEALTH CHECK
# ============================================================

@app.get("/")
def root():

    return {
        "status": "ok",
        "service": "WhatsApp Finance Parser"
    }


# ============================================================
# 4. PARSE TRANSACTION ENDPOINT
# ============================================================

@app.post("/parse-transaction")
def parse_transaction(
    request: TransactionRequest
):

    result = create_database_payload(
        request.message
    )

    return result


# ============================================================
# 5. SHOW ROUTES
# ============================================================

print("=" * 70)
print("FASTAPI APPLICATION CREATED")
print("=" * 70)

print("Available endpoints:")

print("GET  /")
print("POST /parse-transaction")
```
